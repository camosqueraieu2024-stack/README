```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("https://raw.githubusercontent.com/manoelgadi/PSDMA/refs/heads/main/datasets/UN.csv")
numeric_cols = df.select_dtypes(include="number").columns.tolist()

# Missing values → column median
X = df[numeric_cols].fillna(df[numeric_cols].median())

# Scaling
scaler = StandardScaler()
Z = scaler.fit_transform(X)

pd.DataFrame(Z, columns=numeric_cols).describe().T[["mean", "std"]].round(2)
```

K-Means groups countries by Euclidean distance, so variables with larger numeric ranges dominate the result. Here, GDP per capita ranges up to about 42,000 US$, while fertility only ranges from 1 to 8. Without scaling, the clusters would essentially be a GDP ranking. StandardScaler converts each indicator to mean 0 and standard deviation 1, so all 12 indicators contribute equally to the clustering.

```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

ks = range(1, 11)
inertia = []

for k in ks:
    km = KMeans(n_clusters=k, n_init=20, random_state=42)
    km.fit(Z)
    inertia.append(km.inertia_)   # within-cluster sum of squares

# Share of total variance explained by each k
total_ss = inertia[0]
for k, wcss in zip(ks, inertia):
    print(f"k = {k:2d} | inertia = {wcss:7.1f} | variance explained = {1 - wcss/total_ss:.1%}")
```

```python
from sklearn.cluster import KMeans
import numpy as np

ks = list(range(1, 11))
inertia = [KMeans(n_clusters=k, n_init=20, random_state=42).fit(Z).inertia_ for k in ks]


x = (np.array(ks) - ks[0]) / (ks[-1] - ks[0])
y = (np.array(inertia) - inertia[-1]) / (inertia[0] - inertia[-1])
distance = (1 - x) - y

k_opt = ks[np.argmax(distance)]
print("Recommended number of clusters (Elbow Rule): k =", k_opt)
```

```python
plt.figure(figsize=(7, 4))
plt.plot(list(ks), inertia, "o-")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia (within-cluster sum of squares)")
plt.title("Elbow Rule")
plt.xticks(list(ks))
plt.grid(alpha=0.3)
plt.show()
```

```python
from sklearn.cluster import KMeans

km2 = KMeans(n_clusters=2, n_init=20, random_state=42).fit(Z)
km_opt = KMeans(n_clusters=k_opt, n_init=20, random_state=42).fit(Z)

df["cluster_k2"] = km2.labels_
df["cluster_opt"] = km_opt.labels_

# How much of the structure each solution explains
total_ss = KMeans(n_clusters=1, n_init=1).fit(Z).inertia_
for name, km in [("k = 2", km2), (f"k = {k_opt}", km_opt)]:
    print(f"{name}: variance explained = {1 - km.inertia_ / total_ss:.1%}, "
          f"cluster sizes = {pd.Series(km.labels_).value_counts().sort_index().to_dict()}")
```

```python
key = ["GDPperCapita", "lifeFemale", "infantMortality", "tfr", "illiteracyFemale"]

print("Median profile, k = 2")
display(df.groupby("cluster_k2")[key].median().round(1))

print(f"Median profile, k = {k_opt}")
display(df.groupby("cluster_opt")[key].median().round(1))
```

```python
# Where do the k = 2 clusters go when we move to k = 3?
display(pd.crosstab(df["cluster_k2"], df["cluster_opt"],
                    rownames=["k = 2"], colnames=[f"k = {k_opt}"]))

for c in sorted(df["cluster_opt"].unique()):
    members = df.loc[df["cluster_opt"] == c, "country"]
    print(f"\nk = {k_opt}, cluster {c} ({len(members)} countries):")
    print(", ".join(members))
```

Conclusion
 Low income: Afghanistan

**Why it belongs**: Afghanistan has some of the worst indicators in the dataset: 6.9 children per woman, female life expectancy of only 46 years, infant mortality of 154 per 1,000 (one of the highest of all 207 countries), and 85% of women illiterate.
Similar to the others? On health and fertility, yes. It matches Mozambique, Niger and Sierra Leone. But it differs on two indicators. Only 7% of women are economically active, against a cluster median of 49%, because in most Level 1 countries women work in subsistence farming. And its reported GDP per capita ($2,848) is about eight times the cluster median ($355), which is likely a data-quality issue.
What distinguishes the cluster: the highest fertility and infant mortality, and the lowest life expectancy, literacy and income.
Surprising: Afghanistan is one of only eight Asian countries in a cluster of 42 African ones. Its health indicators are so extreme that they outweigh its unusual GDP and female employment.

MIddle Argentina

**Why it belongs **: Argentina has fertility of 2.6, female life expectancy of 77, infant mortality of 22 and female illiteracy below 4%. Its social outcomes are close to rich countries, with a GDP per capita of $8,055.
Similar to the others? Yes. It closely resembles Uruguay, Chile, Mexico and Brazil, as well as Eastern European countries such as Poland and Hungary.
What distinguishes the cluster: near-replacement fertility, low infant mortality and high literacy, but incomes far below the rich group.
Surprising: Argentina is one of the richest members of this cluster, at almost the same income as Greece ($8,684), which is also here. It stays out of Level 4 because its infant mortality (22) is more than three times the rich-country level (about 6).

 High income: Australia

**Why it belongs**: Australia has GDP per capita of $20,046, female life expectancy of 81, infant mortality of 6, fertility of 1.9, and 76% contraception use.
Similar to the others? Very much. It is one of the most typical members, sitting just behind Canada, alongside New Zealand, the USA and Western Europe.
What distinguishes the cluster: the highest income and life expectancy, and the lowest fertility and infant mortality.
Surprising: Australia shares this cluster with French overseas territories such as Réunion and Martinique, which are geographically in the Global South but have health indicators shaped by the French system.
